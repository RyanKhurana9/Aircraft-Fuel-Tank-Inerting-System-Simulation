```python
plt.figure(figsize=(10, 6))

plt.plot(df["fuel_mass_kg"], df["o2_fraction"])

plt.xlabel("Fuel Mass (kg)")
plt.ylabel("Oxygen Fraction")
plt.title("Fuel Mass vs Oxygen Concentration")

plt.show()
```

```python
plt.plot(df['altitude_m'],df['tank_temperature_K'])

plt.xlabel('Altitude (m)')
plt.ylabel('Tank Temperature (K)')
plt.title('Tank Temperature vs Altitude')
plt.grid()
```

```python
plt.plot(df['fuel_mass_kg'],df['ullage_volume_m3'])
plt.xlabel('Fuel Mass (kg)')
plt.ylabel('Ullage Volume (m^3)')
plt.title('Ullage Volume vs Fuel Mass')
plt.grid()
```
